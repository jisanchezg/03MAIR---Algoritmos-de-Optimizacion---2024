#Modelo

**¿Cómo represento el espacio de soluciones?**

El espacio de soluciones se está representando mediante un array de longitud igual al número de tomas (30), donde la posición t (correspondiente a la toma t) almacena un valor entero que indica el día en el que se asigna dicha toma.

Esto podría expresarlo diciendo que cada solución es un vector x=(x<sub>1</sub>​,x<sub>2​</sub>,…,x<sub>30</sub>​) en el que x<sub>t</sub>​ es un día (entre 0 y 4, por ejemplo, si usamos 5 días) tal que no se exceda el máximo de 6 tomas por día.

De esta forma, toda la información sobre qué tomas se graban en cada día queda codificada en este vector de 30 componentes. El resto de variables derivadas (por ejemplo, cuántos actores y cuántas tomas se concentran en cada día) se pueden calcular leyendo la asignación.<br>
<br>

**¿Cual es la función objetivo?**

La función objetivo es minimizar la suma total de actores distintos que participan cada día. Dicho de otro modo, cada día en el que se graban varias tomas, contamos el número de actores distintos que participan en esas tomas, y sumamos ese recuento para todos los días.

Buscamos la asignación de tomas a días que minimice esa suma.

En términos formales, si D es el conjunto de días y Asignar(d) las tomas grabadas el día d, entonces:

$$
\text{Coste} \;=\; \sum_{d \in D}
  \Bigl|\;\bigcup_{t \in \text{Asignar}(d)} \text{ActoresNecesarios}(t)\Bigr|
$$

Ese valor es precisamente el número de "actor-días" totales, que corresponde al gasto por pagar a cada actor presente en cada jornada.<br>
<br>

**¿Cómo implemento las restricciones?**

En esta aproximación heurística (voraz + búsqueda local), las restricciones se manejan de forma implícita en el propio código:

1. Durante la asignación (en el algoritmo voraz) comprobamos que si un día dd ya tiene 6 tomas, no se asignan más tomas a ese día. De esta forma, nunca se supera la cuota de 6 tomas diarias.

2. Cada toma se asigna a exactamente un día.

    - La representación de la solución (lista de 30 posiciones, cada una con un día) ya fuerza que cada toma tt tenga exactamente un día asignado, p. ej. asignacion[t] = d.

    - En el algoritmo voraz, cuando procesamos la toma tt, elegimos un día válido y la asignamos allí.
    
    - En la búsqueda local, cualquier cambio (mover la toma tt a otro día o intercambiar tomas) verifica también la restricción de no exceder las 6 tomas en el día de destino.

3. Todas las tomas deben asignarse.

   - Nuestro algoritmo recorre las 30 tomas (por ejemplo, en orden), asignando cada una a un día válido. Al final del proceso, cada toma ha quedado asignada a exactamente un día.

#Análisis
- ¿Que complejidad tiene el problema?. Orden de complejidad y Contabilizar el espacio de soluciones

Asumiendo que debemos repartir exactamente 30 tomas en 5 días, con 6 tomas al día sin importar el orden dentro de cada día, la cantidad de formas de agruparlas es:

$$
\frac{30!}{(6!)^5} \approx 3.24 \times 10^{23}
$$

Este número ya refleja un espacio de soluciones enorme (del orden de 10<sup>23</sup>), lo cual hace que no sea práctico explorar exhaustivamente todas las asignaciones en un tiempo razonable.

En orden de complejidad, la exploración exhaustiva es exponencial en función del número de tomas y días.

Para un caso como este se pueden utlizar heurísticas, que no garantizan la optimalidad pero encuentran soluciones buenas en tiempo mucho más corto.

- ¿Que técnica utilizo? ¿Por qué?

He decido utilizar una técnica heurística que combina un algoritmo voraz para la construcción inicial de la solución con una búsqueda local para refinar y mejorar dicha solución inicial.

Los motivos que me han llevado a elegir esta aproximación depués de buscar alternativas han sido:

- Simplicidad y rapidez de implementación:

    Un enfoque voraz + búsqueda local es relativamente sencillo de programar sin necesidad de grandes estructuras para ramificación, poda o modelos exactos. Además, no requiere librerías externas de optimización.

- Eficacia en problemas combinatorios:

    Aunque no garantiza la solución óptima, un algoritmo voraz inicial suele ofrecer soluciones de calidad razonable en muy poco tiempo. Luego, la búsqueda local permite refinar la asignación inicial, explorando pequeñas modificaciones (mover una toma de un día a otro, intercambiar tomas, etc.) para reducir el coste total.

- Tamaño moderado del problema:

    Con 30 tomas y 10 actores, el problema sigue siendo lo bastante grande como para descartar una búsqueda exhaustiva pura sin estrategias avanzadas de poda, pero lo bastante pequeño para que un método heurístico funcione de manera muy rápida.

#Implementación

```
import requests
import io
import pandas as pd
import random
import math
from collections import defaultdict

# 1) LECTURA DE DATOS DESDE EXCEL
######################################################

# Cargamos el fichero Excel directamente desde la URL
# Lo he subido a una URL propia ya que no conseguía leerlo desde
# Google Sheets
url_excel = "http://gestoresdecontenidos.com/datos-problema-doblaje.xlsx"

# 1) Descargamos el contenido desde la URL
r = requests.get(url_excel)

# 2) Convertimos a un "stream" de bytes
data = io.BytesIO(r.content)

# 3) Leemos con pandas, especificando engine y parámetros
df = pd.read_excel(
    data,
    engine="openpyxl",   # importante para .xlsx
    skiprows=1,          # omite la primera fila ("Actor" o cabecera)
    usecols="B:K",       # columnas B..K son 10 columnas => Actores
    nrows=30             # 30 tomas
)

# Con df tenemos ya 30 filas × 10 columnas de 0/1
actores_por_toma = df.values.tolist()  # convertimos a lista de listas

num_tomas = 30
num_actores = 10
max_tomas_por_dia = 6
num_dias = 5   # supondremos 5 días (5×6 = 30 tomas)

# Creamos la lista de conjuntos: ActoresNecesarios[t] = { actores que participan en la toma t }
ActoresNecesarios = []
for t in range(num_tomas):
    s = set()
    for a in range(num_actores):
        if actores_por_toma[t][a] == 1:
            s.add(a)
    ActoresNecesarios.append(s)


# 2) FUNCIÓN PARA CALCULAR EL COSTE
######################################################

def calcular_coste(asignacion, ActoresNecesarios, num_dias):
    """
    asignacion[t] = d indica que la toma t va al día d.
    El coste es la suma de (número de actores distintos) por cada día.
    """
    actores_en_dia = [set() for _ in range(num_dias)]
    for t, d in enumerate(asignacion):
        actores_en_dia[d] |= ActoresNecesarios[t]
    return sum(len(actores_en_dia[d]) for d in range(num_dias))


# 3) HEURÍSTICA VORAZ PARA UNA SOLUCIÓN INICIAL
######################################################

def construir_solucion_voraz(num_dias=5):
    """
    Asigna primero las tomas con más actores; para cada toma,
    elige el día donde menos aumente el número de actores.
    """
    # Ordenamos tomas por nº de actores (descendente)
    tomas_ordenadas = sorted(range(num_tomas),
                             key=lambda t: len(ActoresNecesarios[t]),
                             reverse=True)

    asignacion = [-1]*num_tomas
    tomas_en_dia = [0]*num_dias
    actores_en_dia = [set() for _ in range(num_dias)]

    for t in tomas_ordenadas:
        mejor_d = None
        mejor_incremento = math.inf

        for d in range(num_dias):
            if tomas_en_dia[d] < max_tomas_por_dia:
                # Coste actual del día d
                coste_d_actual = len(actores_en_dia[d])
                # Si asignamos la toma t a d
                nuevos_actores = ActoresNecesarios[t] - actores_en_dia[d]
                coste_d_nuevo = coste_d_actual + len(nuevos_actores)
                incremento = coste_d_nuevo - coste_d_actual

                if incremento < mejor_incremento:
                    mejor_incremento = incremento
                    mejor_d = d

        # Asignamos la toma t al día 'mejor_d'
        asignacion[t] = mejor_d
        tomas_en_dia[mejor_d] += 1
        actores_en_dia[mejor_d] |= ActoresNecesarios[t]

    return asignacion


# 4) BÚSQUEDA LOCAL PARA MEJORAR LA SOLUCIÓN
######################################################

def busqueda_local(asign_inicial, ActoresNecesarios, num_dias=5,
                   max_iter=1000):
    """
    Tomamos la asignación inicial y repetimos movimientos aleatorios
    (mover una toma de un día a otro) siempre que mejore la solución.
    """
    asignacion = asign_inicial[:]
    mejor_coste = calcular_coste(asignacion, ActoresNecesarios, num_dias)

    iter_sin_mejora = 0

    while iter_sin_mejora < max_iter:
        # Escogemos una toma aleatoriamente
        t = random.randrange(num_tomas)
        d_actual = asignacion[t]
        # Escogemos un día nuevo al azar
        d_nuevo = random.randrange(num_dias)

        if d_nuevo != d_actual:
            # Verificamos cuántas tomas hay en d_nuevo
            tomas_d_nuevo = [tt for tt in range(num_tomas)
                             if asignacion[tt] == d_nuevo]
            if len(tomas_d_nuevo) < max_tomas_por_dia:
                # Hacemos el movimiento
                asignacion[t] = d_nuevo

                cost_nuevo = calcular_coste(asignacion, ActoresNecesarios, num_dias)
                if cost_nuevo < mejor_coste:
                    mejor_coste = cost_nuevo
                    iter_sin_mejora = 0
                else:
                    # revertimos
                    asignacion[t] = d_actual
                    iter_sin_mejora += 1
            else:
                iter_sin_mejora += 1
        else:
            iter_sin_mejora += 1

    return asignacion, mejor_coste


# 5) EJECUCIÓN DE LA HEURÍSTICA
######################################################

# Construcción voraz
sol_inicial = construir_solucion_voraz(num_dias)
coste_inicial = calcular_coste(sol_inicial, ActoresNecesarios, num_dias)
print("Coste de la solución inicial (voraz):", coste_inicial)

# Búsqueda local
sol_final, coste_final = busqueda_local(sol_inicial, ActoresNecesarios, num_dias, max_iter=2000)
print("Coste tras búsqueda local:", coste_final)


# 6) MOSTRAR RESULTADOS
######################################################

dias = defaultdict(list)
for toma, dia in enumerate(sol_final):
    dias[dia].append(toma)

print("\nASIGNACIÓN DE TOMAS A DÍAS:\n")
for d in sorted(dias.keys()):
    print(f"  Día {d}: tomas {dias[d]}")
    actores_d = set()
    for t in dias[d]:
        actores_d |= ActoresNecesarios[t]
    print(f"     Actores presentes: {actores_d}  (total {len(actores_d)})")

print(f"\nCoste total = {coste_final}")
```
